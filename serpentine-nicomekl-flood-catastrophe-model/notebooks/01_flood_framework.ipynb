```python
RUN_NEW_EVENTS = False
if RUN_NEW_EVENTS:
    subprocess.run([sys.executable, str(ROOT / "src/run_simulation.py"), "--n-sims", "10", "--seed", "42", "--output-dir", str(ROOT / "outputs/example")], cwd=ROOT, check=True)
```
